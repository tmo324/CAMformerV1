```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

# Get all CSV files in the directory
csv_files = sorted(glob.glob('/content/cam_1x1_tr0_*.csv'))

# Create figure and axis
plt.figure(figsize=(3*0.9, 2*0.9), dpi=600)

# Read and plot each file
for file in csv_files:
    # Read the CSV file
    df = pd.read_csv(file)

    # Extract trial number from filename for legend
    trial_num = file.split('_')[3].split('.')[0]

    # Plot TIME vs v_ml1
    plt.plot(df['TIME'], df[' v_ml_indiv1_1'], label=f'Trial {trial_num}')

# Customize the plot
plt.xlabel('Time')
plt.ylabel('Matchline Voltage')
# plt.title('v_ml1 vs Time for All Trials')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)

# Adjust layout to prevent legend cutoff
plt.tight_layout()

# Save the plot
plt.savefig('combined_trials_plot.png', bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

# Get all CSV files in the directory and sort them in reverse order
# This ensures Trial 10 is plotted first and appears at the top of the legend
csv_files = sorted(glob.glob('/content/cam_1x1_tr0_*.csv'))[::-1]

# Create figure and axis with consistent sizing and DPI
fig, ax = plt.subplots(figsize=(3*0.9, 2*0.9), dpi=600)

# Read and plot each file
for file in csv_files:
    # Read the CSV file
    df = pd.read_csv(file)

    # Extract trial number from filename for legend
    trial_num = file.split('_')[3].split('.')[0]

    # Plot TIME vs v_ml1 with consistent styling
    ax.plot(df['TIME'], df[' v_ml_indiv1_1'],
            label=f'{trial_num}/10',
            # color='black',
            linewidth=2)
#
# Customize the plot with consistent formatting
ax.set_xlabel('Time (s)', fontsize=7)
ax.set_ylabel('Matchline Voltage', fontsize=7)
ax.tick_params(axis='both', labelsize=7)

# Add grid with consistent style
ax.grid(True, which="both", ls="-", alpha=0.2)


# Add legend outside the plot
# bbox_to_anchor positions the legend: (1.02, 1) means slightly right of the plot, aligned with top
ax.legend(fontsize=5,
         frameon=False,
         loc='upper left',
         bbox_to_anchor=(1.02, 1))

# Adjust layout to prevent legend cutoff
plt.tight_layout()

# Save the plot with tight bounds and extra space for legend
plt.savefig('combined_trials_plot.png',
            bbox_inches='tight',
            dpi=600,
            pad_inches=0.1)  # Add small padding

plt.show()
```
